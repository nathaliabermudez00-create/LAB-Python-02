```python
temperaturas = [0, 10, 20, 30, 40]

resultado = list(map(lambda c: (c * 9/5) + 32, temperaturas))

print(resultado)
```

```python
lista1 = [1, 2, 3, 4, 5]
lista2 = [10, 20, 30, 40, 50]

resultado = list(map(lambda a, b: a + b, lista1, lista2))

print(resultado)
```

```python
import math

radios = [1, 2, 3, 4, 5]

resultado = list(map(lambda radio: math.pi * radio**2, radios))

print(resultado)
```

```python
precios = {
    "pan": 1000,
    "leche": 2500,
    "café": 5000
}

resultado = list(map(lambda producto: (producto[0], producto[1] * 1.10), precios.items()))

print(resultado)
```

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

resultado = list(map(lambda fila: list(map(lambda numero: numero * 10, fila)), matriz))

print(resultado)
```
