```python
multiplo_de_3 = lambda numero: numero % 3 == 0

print(multiplo_de_3(9))
print(multiplo_de_3(10))
print(multiplo_de_3(15))
print(multiplo_de_3(7))
```

```python
cubo = lambda numero: numero ** 3

print(cubo(2))
print(cubo(3))
print(cubo(5))
print(cubo(10))
```

```python
producto = lambda num1, num2: num1 * num2

print(producto(3, 4))
print(producto(5, 6))
print(producto(10, 8))
print(producto(7, 9))
```

```python
mayor = lambda num1, num2: num1 if num1 > num2 else num2

print(mayor(5, 8))
print(mayor(10, 3))
print(mayor(7, 7))
print(mayor(-2, 4))
```

```python
empieza_con_a = lambda palabra: palabra.lower().startswith("a")

print(empieza_con_a("Ana"))
print(empieza_con_a("avión"))
print(empieza_con_a("Casa"))
print(empieza_con_a("ARBOL"))
```

```python
celsius_a_fahrenheit = lambda celsius: (celsius * 9/5) + 32

print(celsius_a_fahrenheit(0))
print(celsius_a_fahrenheit(10))
print(celsius_a_fahrenheit(25))
print(celsius_a_fahrenheit(100))
```

```python
funciones = [
    lambda numero: numero * 2,
    lambda numero: numero + 10,
    lambda numero: numero ** 2
]

numero = int(input("Ingrese un número: "))

for funcion in funciones:
    print(funcion(numero))
```
