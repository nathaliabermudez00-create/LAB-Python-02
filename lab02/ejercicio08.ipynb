```python

def minimax(piedras, turno):
    if piedras == 0:
        if turno == "MAX":
            return -1
        else:
            return 1

    if turno == "MAX":
        mejor_resultado = -2

        for cantidad in range(1, min(3, piedras) + 1):
            resultado = minimax(piedras - cantidad, "MIN")

            if resultado > mejor_resultado:
                mejor_resultado = resultado

        return mejor_resultado

    else:
        mejor_resultado = 2

        for cantidad in range(1, min(3, piedras) + 1):
            resultado = minimax(piedras - cantidad, "MAX")

            if resultado < mejor_resultado:
                mejor_resultado = resultado

        return mejor_resultado


def mejor_jugada(piedras):
    mejor_resultado = -2
    mejor_cantidad = 1

    for cantidad in range(1, min(3, piedras) + 1):
        resultado = minimax(piedras - cantidad, "MIN")

        if resultado > mejor_resultado:
            mejor_resultado = resultado
            mejor_cantidad = cantidad

    return mejor_cantidad


def jugar():
    piedras = 10

    print("Bienvenido al juego NIM")
    print("Hay 10 piedras.")
    print("Puedes quitar entre 1 y 3 piedras.")
    print("Tu empiezas.\n")

    while piedras > 0:

        print("Piedras restantes:", piedras)

        cantidad = int(input("¿Cuantas piedras quieres quitar? "))

        while cantidad < 1 or cantidad > 3 or cantidad > piedras:
            print("Movimiento invalido. Intenta de nuevo.")
            cantidad = int(input("¿Cuantas piedras quieres quitar? "))

        piedras = piedras - cantidad

        print("Quitaste", cantidad, "piedras.")

        if piedras == 0:
            print("¡Ganaste! La computadora perdio.")
            break

        print("\nPiedras restantes:", piedras)

        cantidad = mejor_jugada(piedras)

        piedras = piedras - cantidad

        print("La computadora quita", cantidad, "piedras.")

        if piedras == 0:
            print("¡La computadora gano!")
            break

    print("Fin del juego.")


jugar()
1
1

1
1
```
