```python
palabras = ["pera", "manzana", "uva", "banano", "fresa"]

resultado = sorted(palabras)

print(resultado)
```

```python
personas = [
    (3, "Carlos"),
    (1, "Ana"),
    (4, "Pedro"),
    (2, "Maria")
]

resultado = sorted(personas, key=lambda persona: persona[0])

print(resultado)
```

```python
productos = [
    ("Camisa", 30),
    ("Pantalón", 55),
    ("Medias", 10),
    ("Chaqueta", 80)
]

resultado = sorted(productos, key=lambda producto: (-producto[1], producto[0]))

print(resultado)
```

```python
palabras = ["computador", "sol", "perro", "programacion", "casa", "mar"]

resultado = sorted(palabras, key=lambda palabra: len(palabra))

print(resultado)
```

```python
numeros = [7, 2, 9, 4, 1, 6, 3, 8, 5]

resultado = sorted(numeros, key=lambda numero: numero % 2)

print(resultado)
```

```python
palabra = "banana"

frecuencias = {letra: palabra.count(letra) for letra in palabra}

resultado = dict(sorted(frecuencias.items(), key=lambda elemento: (-elemento[1], elemento[0])))

print(resultado)
```

```python
empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnología", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnología", 28)
]

resultado = sorted(empleados, key=lambda empleado: (empleado[1], empleado[2]))

print(resultado)
```
