```python
palabras = ["casa", "computador", "sol", "perro", "programacion", "luz"]

resultado = list(filter(lambda palabra: len(palabra) > 4, palabras))

print(resultado)
```

```python
elementos = [10, None, 20, None, 30, 40, None]

resultado = list(filter(lambda elemento: elemento is not None, elementos))

print(resultado)
```

```python
palabras = ["avion", "casa", "elefante", "perro", "iglesia", "oso", "uva"]

resultado = list(filter(lambda palabra: palabra[0].lower() in "aeiou", palabras))

print(resultado)
```

```python
palabras = ["oso", "casa", "radar", "perro", "reconocer", "sal"]

resultado = list(filter(lambda palabra: palabra == palabra[::-1], palabras))

print(resultado)
```

```python
numeros = [15, 20, 35, 42, 55, 68, 75, 90]

resultado = list(filter(lambda numero: numero % 10 == 5, numeros))

print(resultado)
```

```python
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

resultado = list(filter(lambda producto: producto["precio"] > 100, productos))

print(resultado)
```
