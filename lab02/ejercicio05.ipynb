```python
from functools import reduce

numeros = [2, 3, 4, 5]

resultado = reduce(lambda a, b: a * b, numeros)

print(resultado)
```

```python
from functools import reduce

palabras = ["Hola", " ", "mundo", " ", "Python"]

resultado = reduce(lambda a, b: a + b, palabras)

print(resultado)
```

```python
from functools import reduce

def mayor(a, b):
    if a > b:
        return a
    else:
        return b

numeros = [12, 45, 7, 89, 23, 56]

resultado = reduce(mayor, numeros)

print(resultado)
```
